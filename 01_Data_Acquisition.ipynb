## API CONFIGURATION

### ENTSO-E

```python
API_KEY = "0d8a1980-60c2-4c84-8762-e2decbdd67fd"

client = EntsoePandasClient(api_key=API_KEY)

# Germany/Luxembourg bidding zone
country_code = "DE_LU"

# Time range
start = pd.Timestamp("2022-01-01", tz="Europe/Berlin")
end = pd.Timestamp("2024-12-31", tz="Europe/Berlin")
```

## Load and Save Data

### Prices data

```python
# Downloading Price Data
day_ahead_prices = client.query_day_ahead_prices(
    country_code,
    start=start,
    end=end
)
```

```python
# Convert to DataFrame
df_prices = day_ahead_prices.to_frame(name="day_ahead_price")

print(df_prices.info())
print(df_prices.head())
```

```python
# SAVING RAW PRICE DATA

df_prices.to_parquet(
    r"C:\Users\SherM\OneDrive\Documents\1. Thesis\Dataset\Raw\day_ahead_prices.parquet"
)

print("Day-ahead data saved successfully.")
```
